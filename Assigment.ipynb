```python
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    operator = input("Enter operator (+, -, *, /, //, %): ")

    if operator == "+":
        print("Result =", x + y)

    elif operator == "-":
        print("Result =", x - y)

    elif operator == "*":
        print("Result =", x * y)

    elif operator == "/":
        if y != 0:
            print("Result =", x / y)
        else:
            print("Cannot divide by zero")

    elif operator == "//":
        if y != 0:
            print("Result =", x // y)
        else:
            print("Cannot divide by zero")

    elif operator == "%":
        if y != 0:
            print("Remainder =", x % y)
        else:
            print("Cannot divide by zero")

    else:
        print("Invalid operator")
```
